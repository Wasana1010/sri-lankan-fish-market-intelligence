```python
import pdfplumber
import pandas as pd
import os
import re
from pathlib import Path

# ============================================================
# 1. FOLDER
# ============================================================

folder = Path(r"F:\25.2F HND\ML CW\official_2026_reports")

# ============================================================
# 2. PDF FILES
# ============================================================

pdf_files = sorted(folder.glob("*.pdf"))

print("PDF files found:", len(pdf_files))

for file in pdf_files:
    print("-", file.name)


# ============================================================
# 3. DETERMINE MONTH AND WEEK FROM FILE NAME
# ============================================================

def get_week_info(filename):

    name = filename.lower()

    month = None

    if "march" in name:
        month = 3
    elif "april" in name:
        month = 4
    elif "may" in name:
        month = 5
    elif "june" in name:
        month = 6
    elif "july" in name:
        month = 7

    if month is None:
        return None, None

    # Find week number
    if "1st" in name or "1stweek" in name:
        week = 1
    elif "2nd" in name or "2-week" in name or "2ndweek" in name:
        week = 2
    elif "3rd" in name or "3rdweek" in name:
        week = 3
    elif "4th" in name or "4thweek" in name:
        week = 4
    else:
        week = None

    return month, week


# ============================================================
# 4. CLEAN NUMBER
# ============================================================

def clean_number(value):

    if value is None:
        return None

    value = str(value).strip()

    # Remove spaces inside numbers
    value = value.replace(" ", "")

    # Remove commas
    value = value.replace(",", "")

    # N/A
    if value.upper() in ["N/A", "NA", "-"]:
        return None

    # Remove brackets used for negative percentage
    value = value.replace("(", "-")
    value = value.replace(")", "")

    try:
        return float(value)
    except:
        return None


# ============================================================
# 5. EXTRACT TEXT FROM PDF
# ============================================================

def extract_pdf_text(pdf_path):

    pages_text = []

    with pdfplumber.open(pdf_path) as pdf:

        for page in pdf.pages:

            text = page.extract_text()

            if text:
                pages_text.append(text)

            else:
                pages_text.append("")

    return pages_text


# ============================================================
# 6. TEST ALL PDFs
# ============================================================

for pdf_file in pdf_files:

    print("\n" + "=" * 70)
    print(pdf_file.name)
    print("=" * 70)

    pages = extract_pdf_text(pdf_file)

    print("Pages:", len(pages))

    for i, text in enumerate(pages):

        print("\n--- PAGE", i + 1, "---")

        print(text[:500])
```

```python
import os
import re
import pandas as pd
import pdfplumber

# ============================================================
# PATHS
# ============================================================

input_folder = r"F:\25.2F HND\ML CW\official_2026_reports"

output_folder = r"F:\25.2F HND\ML CW\processed_data"

os.makedirs(output_folder, exist_ok=True)


# ============================================================
# GET MONTH AND WEEK
# ============================================================

def get_period(filename):

    name = filename.lower()

    month = None

    month_names = [
        "january", "february", "march", "april",
        "may", "june", "july", "august",
        "september", "october", "november", "december"
    ]

    for m in month_names:
        if m in name:
            month = m.capitalize()
            break

    # Detect 1st, 2nd, 3rd, 4th
    match = re.search(
        r'([1-4])(?:st|nd|rd|th)?[-_ ]*week',
        name
    )

    if match:
        week = int(match.group(1))
    else:
        week = None

    return month, week


# ============================================================
# CLEAN PRICE
# ============================================================

def clean_price(value):

    if value is None:
        return None

    value = str(value).strip()

    if value.upper() in ["N/A", "NA", "-", ""]:
        return None

    # Fix PDF extraction such as:
    # "2 ,900.00"
    # "9 60.00"
    value = value.replace(" ", "")

    # Remove commas
    value = value.replace(",", "")

    # Remove brackets
    value = value.replace("(", "")
    value = value.replace(")", "")

    # Extract numeric value
    match = re.search(r'\d+(?:\.\d+)?', value)

    if match:
        try:
            return float(match.group())
        except:
            return None

    return None


# ============================================================
# EXTRACT ONE TABLE FROM PDF TEXT
# ============================================================

def extract_rows(text):

    rows = []

    lines = text.split("\n")

    for line in lines:

        line = line.strip()

        # Only process lines beginning with species number
        match = re.match(r'^(\d+)\s+(.*)', line)

        if not match:
            continue

        number = int(match.group(1))
        remaining = match.group(2).strip()

        # Ignore unexpected numbers
        if number < 1 or number > 40:
            continue

        # ----------------------------------------------------
        # Find the first Sinhala/Unicode section and English
        # common name + price values.
        # ----------------------------------------------------

        # Extract all numeric price-like values
        price_matches = re.findall(
            r'(?:\d[\d ,]*\.\d+|\d[\d ]+\.\d+)',
            remaining
        )

        # We need at least one price
        if len(price_matches) == 0:
            continue

        prices = [
            clean_price(x)
            for x in price_matches
        ]

        prices = [
            x for x in prices
            if x is not None
        ]

        if len(prices) < 1:
            continue

        # ----------------------------------------------------
        # Try to identify common name
        # ----------------------------------------------------

        # Find position before first price
        first_price = re.search(
            r'(?:\d[\d ,]*\.\d+|\d[\d ]+\.\d+)',
            remaining
        )

        if first_price:

            name_part = remaining[:first_price.start()].strip()

        else:

            name_part = remaining

        # Remove obvious Sinhala section by taking the
        # final English-looking portion where possible.
        english_parts = re.findall(
            r'[A-Za-z][A-Za-z /()\-]+',
            name_part
        )

        if english_parts:

            common_name = english_parts[-1].strip()

        else:

            common_name = name_part.strip()

        rows.append({
            "Number": number,
            "Common_Name": common_name,
            "Prices": prices
        })

    return rows


# ============================================================
# PROCESS ALL PDFs
# ============================================================

all_data = []

pdf_files = sorted([
    f for f in os.listdir(input_folder)
    if f.lower().endswith(".pdf")
])

print("=" * 70)
print("PROCESSING FISH PRICE REPORTS")
print("=" * 70)
print("PDF files:", len(pdf_files))


for filename in pdf_files:

    file_path = os.path.join(input_folder, filename)

    month, week = get_period(filename)

    print("\nProcessing:", filename)
    print("Month:", month, "| Week:", week)

    try:

        with pdfplumber.open(file_path) as pdf:

            for page_number, page in enumerate(
                pdf.pages,
                start=1
            ):

                text = page.extract_text()

                if not text:
                    continue

                rows = extract_rows(text)

                for row in rows:

                    prices = row["Prices"]

                    all_data.append({
                        "Year": 2026,
                        "Month": month,
                        "Week": week,
                        "Period": f"2026-{month}-W{week}",
                        "Page": page_number,
                        "Number": row["Number"],
                        "Species": row["Common_Name"],
                        "Price_1": prices[0]
                        if len(prices) >= 1 else None,
                        "Price_2": prices[1]
                        if len(prices) >= 2 else None,
                        "Price_3": prices[2]
                        if len(prices) >= 3 else None,
                        "Price_4": prices[3]
                        if len(prices) >= 4 else None,
                        "Price_5": prices[4]
                        if len(prices) >= 5 else None
                    })

        print("Rows extracted:", len(rows))

    except Exception as e:

        print("ERROR:", e)


# ============================================================
# CREATE DATAFRAME
# ============================================================

df_raw = pd.DataFrame(all_data)

print("\n" + "=" * 70)
print("EXTRACTION COMPLETE")
print("=" * 70)

print("Total rows:", len(df_raw))

display(df_raw.head(20))
```

```python
# ============================================================
# MACHINE LEARNING - FISH MARKET PRICE PREDICTION
# SRI LANKA
#
# Extract data from 18 official weekly PDF reports
# Identify Wholesale / Consumer prices
# Clean species names
# Create fish_price_final.csv
# ============================================================

import os
import re
import pdfplumber
import pandas as pd
import numpy as np


# ============================================================
# 1. FOLDER PATH
# ============================================================

folder_path = r"F:\25.2F HND\ML CW\official_2026_reports"

output_file = r"F:\25.2F HND\ML CW\fish_price_final.csv"


print("=" * 80)
print("FISH PRICE DATA EXTRACTION")
print("=" * 80)

print("Input folder:")
print(folder_path)

print("\nFolder exists:", os.path.exists(folder_path))


if not os.path.exists(folder_path):
    raise FileNotFoundError(
        f"Folder not found:\n{folder_path}"
    )


# ============================================================
# 2. FIND ALL PDF FILES
# ============================================================

pdf_files = [
    os.path.join(folder_path, f)
    for f in os.listdir(folder_path)
    if f.lower().endswith(".pdf")
]

pdf_files = sorted(pdf_files)


print("\nPDF files found:", len(pdf_files))

for pdf in pdf_files:
    print(os.path.basename(pdf))


if len(pdf_files) == 0:
    raise FileNotFoundError("No PDF files found.")


# ============================================================
# 3. SPECIES NAME CLEANING
# ============================================================

def clean_species_name(name):

    if name is None:
        return None

    name = str(name)

    # --------------------------------------------------------
    # Remove line breaks
    # --------------------------------------------------------

    name = name.replace("\n", " ")
    name = name.replace("\r", " ")

    # --------------------------------------------------------
    # Remove excessive spaces
    # --------------------------------------------------------

    name = re.sub(r"\s+", " ", name).strip()

    # --------------------------------------------------------
    # Remove Sinhala text
    # Keep English species names
    # --------------------------------------------------------

    # If English text exists after Sinhala text,
    # extract the English portion where possible.
    
    # Common OCR cases
    name = re.sub(r"^[^A-Za-z]*", "", name)

    # --------------------------------------------------------
    # Fix common OCR / PDF extraction errors
    # --------------------------------------------------------

    replacements = {

        # Prawns
        "M) Prawns (M)": "Prawns (M)",
        "M) Prawns": "Prawns (M)",
        "Prawns (M) 3": "Prawns (M)",

        # Sea Crabs
        "L) Sea Crabs(L)": "Sea Crabs (L)",
        "Sea Crabs(L)": "Sea Crabs (L)",

        # Rock fish
        "Rock Fish (L)": "Rock fish (L)",

        # Mackerel
        "Indian mackerel": "Indian Mackerel",

        # Shark
        "Shark": "Sharks",

        # Anchovy
        "Anchovies": "Anchovy",

        # Sardinella
        "Sardinella": "Sardinella",

        # Tilapia
        "Thilapia (M)": "Tilapia (M)",

        # Cuttle fish
        "Cuttle fish": "Squids / Cuttle fish (Peeli)",

        # Rainbow Runner
        "Rainbow runner": "Rainbow Runner",

        # Threadfin
        "Threadfin bream": "Threadfin Bream",

        # Ginnati
        "Ginnati paraw": "Ginnati Paraw",

        # Frigate tuna
        "Frigate tuna": "Frigate tuna",

        # Barracuda
        "Barracuda (L)": "Barracuda (L)",

        # Indian Scad
        "Indian Scade": "Indian Scad",
    }

    if name in replacements:
        name = replacements[name]

    # --------------------------------------------------------
    # Standard spacing around brackets
    # --------------------------------------------------------

    name = name.replace("(L)", " (L)")
    name = name.replace("(S)", " (S)")
    name = name.replace("(M)", " (M)")

    # Fix double spaces
    name = re.sub(r"\s+", " ", name).strip()

    return name


# ============================================================
# 4. STANDARD SPECIES MAPPING
# ============================================================

species_mapping = {

    "Seer (Ni-L)": "Seer (Ni-L)",
    "Seer": "Seer",

    "Trevally (L)": "Trevally (L)",

    "Rock fish (L)": "Rock fish (L)",
    "Rock fish (S)": "Rock fish (S)",

    "Sail fish": "Sail fish",

    "Skipjack tuna": "Skipjack tuna",

    "Yellowfin tuna": "Yellowfin tuna",

    "Sardinella": "Sardinella",

    "Sharks": "Sharks",
    "Shark": "Sharks",

    "Herrings": "Herrings",

    "Indian Mackerel": "Indian Mackerel",
    "Indian mackerel": "Indian Mackerel",

    "Pony fish": "Pony fish",

    "Katuwalla": "Katuwalla",

    "Anchovy": "Anchovy",
    "Anchovies": "Anchovy",

    "Prawns (M)": "Prawns (M)",

    "Marlins": "Marlins",

    "Frigate tuna": "Frigate tuna",
    "Frigate tuna (L)": "Frigate tuna (L)",

    "Kawakawa": "Kawakawa",

    "Red Bream (L)": "Red Bream (L)",
    "Red Bream(L)": "Red Bream (L)",

    "Big eye scade": "Big eye scade",

    "Ginnati Paraw": "Ginnati Paraw",
    "Ginnati paraw": "Ginnati Paraw",

    "Needle fish": "Needle fish",

    "Indian Anchovies": "Indian Anchovies",

    "Barracuda (L)": "Barracuda (L)",
    "Barracuda(L)": "Barracuda (L)",

    "Indian Scad (L)": "Indian Scad (L)",
    "Indian Scad(L)": "Indian Scad (L)",

    "Indian Scad (S)": "Indian Scad (S)",
    "Indian Scad(S)": "Indian Scad (S)",

    "Rainbow Runner": "Rainbow Runner",

    "Threadfin Bream": "Threadfin Bream",

    "White Sardinella": "White Sardinella",

    "Squids / Cuttle fish (Peeli)": "Squids / Cuttle fish (Peeli)",

    "Sea Crabs (L)": "Sea Crabs (L)",

    "Tilapia (M)": "Tilapia (M)",
}


def standardize_species(name):

    name = clean_species_name(name)

    if name in species_mapping:
        return species_mapping[name]

    return name


# ============================================================
# 5. GET YEAR / MONTH / WEEK FROM FILE NAME
# ============================================================

def extract_period_from_filename(filename):

    filename_lower = filename.lower()

    # -----------------------------
    # Year
    # -----------------------------

    year_match = re.search(r"(20\d{2})", filename)

    if year_match:
        year = int(year_match.group(1))
    else:
        year = 2026

    # -----------------------------
    # Month
    # -----------------------------

    months = [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June",
        "July",
        "August",
        "September",
        "October",
        "November",
        "December"
    ]

    month = None

    for m in months:

        if m.lower() in filename_lower:
            month = m
            break

    # -----------------------------
    # Week
    # -----------------------------

    week = None

    patterns = [
        r"(\d+)(?:st|nd|rd|th)[-_ ]?week",
        r"(\d+)[-_ ]?week",
        r"week[-_ ]?(\d+)"
    ]

    for pattern in patterns:

        match = re.search(
            pattern,
            filename_lower
        )

        if match:
            week = int(match.group(1))
            break

    return year, month, week


# ============================================================
# 6. CLEAN NUMERIC VALUES
# ============================================================

def clean_number(value):

    if value is None:
        return np.nan

    value = str(value).strip()

    # Remove spaces
    value = value.replace(" ", "")

    # Remove commas
    value = value.replace(",", "")

    # Remove brackets used for negative percentages
    if value.startswith("(") and value.endswith(")"):

        value = "-" + value[1:-1]

    # Remove unwanted characters except digits, dot and minus

    value = re.sub(
        r"[^0-9.\-]",
        "",
        value
    )

    if value == "":
        return np.nan

    try:
        return float(value)

    except:
        return np.nan


# ============================================================
# 7. IDENTIFY PRICE TABLE
# ============================================================

def identify_table_type(text):

    if text is None:
        return None

    text_lower = text.lower()

    # Wholesale table

    if (
        "wholesale prices" in text_lower
        or "peliyagoda fish market" in text_lower
    ):
        return "Wholesale"

    # Consumer table

    if (
        "consumer prices" in text_lower
        or "selected markets" in text_lower
    ):
        return "Consumer"

    return None


# ============================================================
# 8. EXTRACT ONE PDF
# ============================================================

def extract_pdf(pdf_path):

    records = []

    filename = os.path.basename(pdf_path)

    year, month, week = extract_period_from_filename(filename)

    print("\n" + "=" * 80)
    print("PROCESSING:", filename)
    print("=" * 80)

    try:

        with pdfplumber.open(pdf_path) as pdf:

            print("Pages:", len(pdf.pages))

            for page_number, page in enumerate(
                pdf.pages,
                start=1
            ):

                print(
                    f"\nProcessing page {page_number}"
                )

                # ------------------------------------------------
                # Extract page text
                # ------------------------------------------------

                page_text = page.extract_text()

                if page_text is None:
                    page_text = ""

                # ------------------------------------------------
                # Identify Wholesale / Consumer
                # ------------------------------------------------

                table_type = identify_table_type(
                    page_text
                )

                print(
                    "Table type:",
                    table_type
                )

                if table_type is None:

                    # Fallback based on page number
                    if page_number == 1:
                        table_type = "Wholesale"

                    elif page_number == 2:
                        table_type = "Consumer"

                    print(
                        "Fallback type:",
                        table_type
                    )

                # ------------------------------------------------
                # Extract tables
                # ------------------------------------------------

                tables = page.extract_tables()

                print(
                    "Tables found:",
                    len(tables)
                )

                for table_index, table in enumerate(
                    tables,
                    start=1
                ):

                    if not table:
                        continue

                    print(
                        f"  Table {table_index}: "
                        f"{len(table)} rows"
                    )

                    # ------------------------------------------------
                    # Process rows
                    # ------------------------------------------------

                    for row in table:

                        if row is None:
                            continue

                        # Convert cells to strings

                        row = [
                            "" if cell is None
                            else str(cell).strip()
                            for cell in row
                        ]

                        # Remove empty cells from end

                        while row and row[-1] == "":
                            row.pop()

                        if len(row) < 3:
                            continue

                        # ------------------------------------------------
                        # Combine row for checking
                        # ------------------------------------------------

                        row_text = " ".join(row)

                        row_lower = row_text.lower()

                        # ------------------------------------------------
                        # Skip headers
                        # ------------------------------------------------

                        if (
                            "variety" in row_lower
                            or "common name" in row_lower
                            or "percentage change" in row_lower
                            or "abbreviations" in row_lower
                        ):
                            continue

                        # ------------------------------------------------
                        # Find species
                        # ------------------------------------------------

                        species = None

                        # In your extracted structure,
                        # species is normally column 3 / index 2.
                        #
                        # Example:
                        # [1, Sinhala, Seer, 1420, 1691, 2066, 22.2, 45.5]

                        possible_species_columns = [
                            2,
                            1,
                            3
                        ]

                        for idx in possible_species_columns:

                            if idx >= len(row):
                                continue

                            candidate = row[idx]

                            candidate = clean_species_name(
                                candidate
                            )

                            if candidate is None:
                                continue

                            # Must contain alphabetic characters

                            if not re.search(
                                r"[A-Za-z]",
                                candidate
                            ):
                                continue

                            # Ignore obvious header text

                            if candidate.lower() in [
                                "variety",
                                "common name",
                                "percentage change"
                            ]:
                                continue

                            species = candidate
                            break

                        if species is None:
                            continue

                        species = standardize_species(
                            species
                        )

                        # ------------------------------------------------
                        # Ignore non-species rows
                        # ------------------------------------------------

                        invalid_species = [
                            "Variety",
                            "Common Name",
                            "Percentage Change",
                            "Abbreviations",
                            "Last week",
                            "Last Year"
                        ]

                        if species in invalid_species:
                            continue

                        # ------------------------------------------------
                        # Find numeric values
                        # ------------------------------------------------

                        numeric_values = []

                        for cell in row:

                            number = clean_number(
                                cell
                            )

                            if not np.isnan(number):
                                numeric_values.append(
                                    number
                                )

                        # Need at least 3 numeric values
                        # for Last Week / Last Year / Current

                        if len(numeric_values) < 3:
                            continue

                        # ------------------------------------------------
                        # Determine price columns
                        # ------------------------------------------------

                        # Normally:
                        #
                        # Price 1 = Previous/current historical price
                        # Price 2 = Last week / previous period
                        # Price 3 = Current week
                        # Price 4 = % change from last week
                        # Price 5 = % change from last year
                        #
                        # Because row number and Sinhala columns may
                        # vary, take the last 5 numeric values when
                        # available.

                        if len(numeric_values) >= 5:

                            values = numeric_values[-5:]

                            price_1 = values[0]
                            price_2 = values[1]
                            current_price = values[2]

                            pct_change_last_week = values[3]
                            pct_change_last_year = values[4]

                        elif len(numeric_values) == 4:

                            values = numeric_values[-4:]

                            price_1 = values[0]
                            price_2 = values[1]
                            current_price = values[2]

                            pct_change_last_week = values[3]
                            pct_change_last_year = np.nan

                        else:

                            price_1 = numeric_values[-3]
                            price_2 = np.nan
                            current_price = numeric_values[-2]

                            pct_change_last_week = np.nan
                            pct_change_last_year = np.nan

                        # ------------------------------------------------
                        # Create record
                        # ------------------------------------------------

                        record = {

                            "Year": year,

                            "Month": month,

                            "Week": week,

                            "Period": (
                                f"{year}-"
                                f"{month}-W{week}"
                                if month is not None
                                and week is not None
                                else None
                            ),

                            "PageNumber": page_number,

                            "MarketType": table_type,

                            "Species": species,

                            "Price_1": price_1,

                            "Price_2": price_2,

                            "Current_Price": current_price,

                            "Pct_Change_Last_Week":
                                pct_change_last_week,

                            "Pct_Change_Last_Year":
                                pct_change_last_year,

                            "Source_File":
                                filename
                        }

                        records.append(
                            record
                        )

    except Exception as e:

        print(
            "\nERROR processing file:",
            filename
        )

        print(
            str(e)
        )

    return records


# ============================================================
# 9. PROCESS ALL PDFs
# ============================================================

all_records = []


print("\n")
print("=" * 80)
print("STARTING EXTRACTION")
print("=" * 80)


for pdf_file in pdf_files:

    records = extract_pdf(
        pdf_file
    )

    all_records.extend(
        records
    )

    print(
        "\nRecords extracted from file:",
        len(records)
    )


# ============================================================
# 10. CREATE DATAFRAME
# ============================================================

df = pd.DataFrame(
    all_records
)


print("\n")
print("=" * 80)
print("EXTRACTION COMPLETED")
print("=" * 80)

print(
    "Total records:",
    len(df)
)


# ============================================================
# 11. REMOVE DUPLICATES
# ============================================================

if not df.empty:

    df = df.drop_duplicates(
        subset=[
            "Year",
            "Month",
            "Week",
            "PageNumber",
            "MarketType",
            "Species"
        ],
        keep="first"
    )


# ============================================================
# 12. CLEAN SPECIES AGAIN
# ============================================================

if not df.empty:

    df["Species"] = (
        df["Species"]
        .apply(standardize_species)
    )


# ============================================================
# 13. CLEAN MARKET TYPE
# ============================================================

if not df.empty:

    df["MarketType"] = (
        df["MarketType"]
        .astype(str)
        .str.strip()
        .str.title()
    )


# ============================================================
# 14. REMOVE INVALID SPECIES
# ============================================================

invalid_words = [

    "Variety",
    "Common Name",
    "Percentage Change",
    "Last Week",
    "Last Year",
    "Abbreviations"
]


if not df.empty:

    for word in invalid_words:

        df = df[
            ~df["Species"]
            .str.contains(
                word,
                case=False,
                na=False
            )
        ]


# ============================================================
# 15. NUMERIC COLUMNS
# ============================================================

numeric_columns = [

    "Price_1",
    "Price_2",
    "Current_Price",
    "Pct_Change_Last_Week",
    "Pct_Change_Last_Year"
]


for column in numeric_columns:

    if column in df.columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )


# ============================================================
# 16. REMOVE IMPOSSIBLE PRICES
# ============================================================

if not df.empty:

    df.loc[
        df["Current_Price"] <= 0,
        "Current_Price"
    ] = np.nan


# ============================================================
# 17. SORT DATA
# ============================================================

if not df.empty:

    month_order = {

        "January": 1,
        "February": 2,
        "March": 3,
        "April": 4,
        "May": 5,
        "June": 6,
        "July": 7,
        "August": 8,
        "September": 9,
        "October": 10,
        "November": 11,
        "December": 12
    }

    df["MonthNumber"] = (
        df["Month"]
        .map(month_order)
    )

    df = df.sort_values(
        by=[
            "Year",
            "MonthNumber",
            "Week",
            "MarketType",
            "Species"
        ]
    )

    df = df.drop(
        columns=["MonthNumber"]
    )


# ============================================================
# 18. RESET INDEX
# ============================================================

df = df.reset_index(
    drop=True
)


# ============================================================
# 19. DISPLAY BASIC INFORMATION
# ============================================================

print("\n")
print("=" * 80)
print("FINAL DATASET INFORMATION")
print("=" * 80)

print(
    "Rows:",
    len(df)
)

print(
    "Columns:",
    len(df.columns)
)


print("\nColumns:")
print(
    df.columns.tolist()
)


# ============================================================
# 20. MARKET TYPE COUNTS
# ============================================================

if not df.empty:

    print("\n")
    print("=" * 80)
    print("MARKET TYPE COUNTS")
    print("=" * 80)

    print(
        df["MarketType"]
        .value_counts()
    )


# ============================================================
# 21. SPECIES COUNT
# ============================================================

if not df.empty:

    print("\n")
    print("=" * 80)
    print("UNIQUE SPECIES")
    print("=" * 80)

    print(
        "Number of species:",
        df["Species"].nunique()
    )

    print("\nSpecies list:")

    for species in sorted(
        df["Species"]
        .dropna()
        .unique()
    ):

        print(
            "-",
            species
        )


# ============================================================
# 22. MISSING VALUE REPORT
# ============================================================

print("\n")
print("=" * 80)
print("MISSING VALUES")
print("=" * 80)

print(
    df.isnull()
    .sum()
)


# ============================================================
# 23. SAVE FINAL CSV
# ============================================================

df.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)


print("\n")
print("=" * 80)
print("CSV CREATED SUCCESSFULLY")
print("=" * 80)

print(
    output_file
)


# ============================================================
# 24. SHOW FIRST 30 ROWS
# ============================================================

print("\n")
print("=" * 80)
print("FIRST 30 ROWS")
print("=" * 80)

display(
    df.head(30)
)
```

```python
import os
import re
import pandas as pd
import pdfplumber

# ============================================================
# 1. SETTINGS
# ============================================================

folder_path = r"F:\25.2F HND\ML CW\official_2026_reports"

output_file = r"F:\25.2F HND\ML CW\fish_price_final.csv"


# ============================================================
# 2. FIND PDF FILES
# ============================================================

pdf_files = sorted([
    os.path.join(folder_path, f)
    for f in os.listdir(folder_path)
    if f.lower().endswith(".pdf")
])

print("=" * 80)
print("PDF FILES")
print("=" * 80)
print("PDF count:", len(pdf_files))

for f in pdf_files:
    print(os.path.basename(f))


# ============================================================
# 3. SPECIES STANDARDIZATION
# ============================================================

def clean_species(species):

    species = str(species).strip()

    # Remove extra spaces
    species = re.sub(r"\s+", " ", species)

    # Fix extraction errors
    species = species.replace("M) Prawns (M)", "Prawns (M)")
    species = species.replace("L) Sea Crabs(L)", "Sea Crabs(L)")
    species = species.replace("R)ock fish (L)", "Rock fish (L)")

    mapping = {

        "seer": "Seer",
        "seer (ni-l)": "Seer (Ni-L)",

        "trevally (l)": "Trevally (L)",

        "rock fish (l)": "Rock fish (L)",
        "rock fish (s)": "Rock fish (S)",

        "sail fish": "Sail fish",

        "skipjack tuna": "Skipjack tuna",

        "yellowfin tuna": "Yellowfin tuna",

        "sardinella": "Sardinella",

        "shark": "Shark",
        "sharks": "Shark",

        "herrings": "Herrings",
        "herring": "Herrings",

        "indian mackerel": "Indian Mackerel",

        "pony fish": "Pony fish",

        "katuwalla": "Katuwalla",

        "anchovy": "Anchovy",
        "anchovies": "Anchovy",

        "prawns (m)": "Prawns (M)",
        'prawns (m) 3"': "Prawns (M)",

        "marlins": "Marlins",

        "frigate tuna": "Frigate tuna(L)",
        "frigate tuna(l)": "Frigate tuna(L)",

        "kawakawa": "Kawakawa",

        "red bream(l)": "Red Bream(L)",

        "big eye scade": "Big eye scade",

        "ginnati paraw": "Ginnati Paraw",

        "needle fish": "Needle fish",

        "indian anchovies": "Indian Anchovies",
        "indian anchovy": "Indian Anchovies",

        "barracuda(l)": "Barracuda(L)",

        "indian scad(l)": "Indian Scad(L)",
        "indian scad(s)": "Indian Scad(S)",
        "indian scade": "Indian Scad(L)",

        "rainbow runner": "Rainbow Runner",

        "threadfin bream": "Threadfin Bream",

        "white sardinella": "White Sardinella",

        "squids /cuttle fish(peeli)": "Squids /Cuttle fish(Peeli)",
        "cuttle fish": "Cuttle fish",

        "sea crabs(l)": "Sea Crabs(L)",

        "tilapia (m)": "Tilapia (M)",
        "thilapia (m)": "Tilapia (M)"
    }

    return mapping.get(species.lower(), species)


# ============================================================
# 4. SPECIES LIST
# ============================================================

species_patterns = [
    r"Seer(?:\s*\(Ni-L\))?",
    r"Trevally\s*\(L\)",
    r"Rock\s*fish\s*\([LS]\)",
    r"Sail\s*fish",
    r"Skipjack\s*tuna",
    r"Yellowfin\s*tuna",
    r"Sardinella",
    r"Sharks?",
    r"Herrings?",
    r"Indian\s*[Mm]ackerel",
    r"Pony\s*fish",
    r"Katuwalla",
    r"Anchovies?",
    r"Prawns\s*\(M\)",
    r"Marlins",
    r"Frigate\s*tuna(?:\(L\))?",
    r"Kawakawa",
    r"Red\s*Bream\s*\(L\)",
    r"Big\s*eye\s*scade",
    r"Ginnati\s*[Pp]araw",
    r"Needle\s*fish",
    r"Indian\s*Anchovies",
    r"Barracuda\s*\(L\)",
    r"Indian\s*Scad\s*\([LS]\)",
    r"Indian\s*Scade",
    r"Rainbow\s*[Rr]unner",
    r"Threadfin\s*[Bb]ream",
    r"White\s*Sardinella",
    r"Squids?\s*/?\s*Cuttle\s*fish(?:\(Peeli\))?",
    r"Cuttle\s*fish",
    r"Sea\s*Crabs?\s*\(L\)",
    r"Tilapia\s*\(M\)",
    r"Thilapia\s*\(M\)"
]

species_regex = re.compile(
    "(" + "|".join(species_patterns) + ")",
    re.IGNORECASE
)


# ============================================================
# 5. CLEAN NUMERIC TEXT
# ============================================================

def parse_number(value):

    if value is None:
        return None

    value = str(value).strip()

    # N/A or -
    if re.search(r"N/?A", value, re.IGNORECASE):
        return None

    if value == "-":
        return None

    # Remove spaces
    value = value.replace(" ", "")

    # Remove commas
    value = value.replace(",", "")

    # Percentage negative numbers in brackets
    if value.startswith("(") and value.endswith(")"):
        value = "-" + value[1:-1]

    # Remove unwanted characters
    value = re.sub(r"[^0-9.\-]", "", value)

    if value == "":
        return None

    try:
        return float(value)
    except:
        return None


# ============================================================
# 6. EXTRACT MONTH AND WEEK
# ============================================================

def get_month_week(filename):

    name = filename.lower()

    months = [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June",
        "July",
        "August",
        "September",
        "October",
        "November",
        "December"
    ]

    month = None

    for m in months:

        if m.lower() in name:
            month = m
            break

    week = None

    if re.search(r"1st[-_ ]?week", name):
        week = 1

    elif re.search(r"2nd[-_ ]?week", name):
        week = 2

    elif re.search(r"3rd[-_ ]?week", name):
        week = 3

    elif re.search(r"4th[-_ ]?week", name):
        week = 4

    return month, week


# ============================================================
# 7. EXTRACT ONE ROW FROM PDF TEXT
# ============================================================

def extract_species_rows(text, market_type):

    rows = []

    lines = text.splitlines()

    for line in lines:

        line = line.strip()

        if not line:
            continue

        # Ignore headers
        if "Common Name" in line:
            continue

        if "Sinhala Name" in line:
            continue

        if "Abbreviations" in line:
            continue

        # Must begin with a row number
        match_number = re.match(r"^(\d+)\s+", line)

        if not match_number:
            continue

        row_number = int(match_number.group(1))

        # --------------------------------------------------------
        # Find species
        # --------------------------------------------------------

        species_match = species_regex.search(line)

        if not species_match:
            continue

        species_raw = species_match.group(0)

        species = clean_species(species_raw)

        # --------------------------------------------------------
        # Text after species
        # --------------------------------------------------------

        after_species = line[species_match.end():]

        # --------------------------------------------------------
        # Extract numeric values
        # --------------------------------------------------------

        # Capture:
        # 1,420.00
        # 650.00
        # 22.2
        # (10.4)
        # -
        # N/A

        number_pattern = re.compile(
            r"(?:\(?\d[\d,\s]*\.?\d*\)?|N/?A|-)"
        )

        raw_values = number_pattern.findall(after_species)

        values = []

        for value in raw_values:

            parsed = parse_number(value)

            values.append(parsed)

        # --------------------------------------------------------
        # Remove weird empty values at beginning
        # --------------------------------------------------------

        while values and values[0] is None:
            values.pop(0)

        # --------------------------------------------------------
        # Need minimum 3 prices
        # --------------------------------------------------------

        if len(values) < 3:
            continue

        price_2025 = values[0]
        price_last_week = values[1]
        price_current = values[2]

        change_last_week = (
            values[3] if len(values) > 3 else None
        )

        change_last_year = (
            values[4] if len(values) > 4 else None
        )

        rows.append({
            "RowNumber": row_number,
            "MarketType": market_type,
            "Species": species,
            "Price_2025": price_2025,
            "Price_LastWeek": price_last_week,
            "Price_Current": price_current,
            "Change_LastWeek_%": change_last_week,
            "Change_LastYear_%": change_last_year
        })

    return rows


# ============================================================
# 8. PROCESS ALL 18 PDFs
# ============================================================

all_data = []

print("\n")
print("=" * 80)
print("STARTING EXTRACTION")
print("=" * 80)


for pdf_path in pdf_files:

    filename = os.path.basename(pdf_path)

    month, week = get_month_week(filename)

    print("\n")
    print("-" * 80)
    print(filename)
    print("Month:", month, "| Week:", week)
    print("-" * 80)

    try:

        with pdfplumber.open(pdf_path) as pdf:

            for page_number, page in enumerate(
                pdf.pages,
                start=1
            ):

                text = page.extract_text()

                if not text:
                    print(
                        "Page",
                        page_number,
                        ": NO TEXT"
                    )
                    continue

                # ------------------------------------------------
                # Identify market type
                # ------------------------------------------------

                text_lower = text.lower()

                if "wholesale" in text_lower:
                    market_type = "Wholesale"

                elif "consumer" in text_lower:
                    market_type = "Consumer"

                else:

                    if page_number == 1:
                        market_type = "Wholesale"

                    elif page_number == 2:
                        market_type = "Consumer"

                    else:
                        market_type = None

                # ------------------------------------------------
                # Extract rows
                # ------------------------------------------------

                page_rows = extract_species_rows(
                    text,
                    market_type
                )

                print(
                    "Page",
                    page_number,
                    "|",
                    market_type,
                    "| Rows:",
                    len(page_rows)
                )

                # ------------------------------------------------
                # Add metadata
                # ------------------------------------------------

                for row in page_rows:

                    row["Year"] = 2026
                    row["Month"] = month
                    row["Week"] = week

                    row["Period"] = (
                        f"2026-{month}-W{week}"
                        if month and week
                        else None
                    )

                    row["PageNumber"] = page_number
                    row["SourceFile"] = filename

                    all_data.append(row)

    except Exception as e:

        print("ERROR:")
        print(e)


# ============================================================
# 9. CREATE DATAFRAME
# ============================================================

df = pd.DataFrame(all_data)

print("\n")
print("=" * 80)
print("EXTRACTION RESULT")
print("=" * 80)

print("Total rows:", len(df))

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 10. STOP IF NOTHING WAS EXTRACTED
# ============================================================

if len(df) == 0:

    print("\n")
    print("ERROR: ZERO ROWS EXTRACTED.")
    print("Do not continue to the next step.")
    print("We need to inspect the PDF text format.")

else:

    # ========================================================
    # 11. REORDER COLUMNS
    # ========================================================

    column_order = [
        "Year",
        "Month",
        "Week",
        "Period",
        "PageNumber",
        "MarketType",
        "Species",
        "Price_2025",
        "Price_LastWeek",
        "Price_Current",
        "Change_LastWeek_%",
        "Change_LastYear_%",
        "RowNumber",
        "SourceFile"
    ]

    df = df[column_order]


    # ========================================================
    # 12. REMOVE DUPLICATES
    # ========================================================

    df = df.drop_duplicates(
        subset=[
            "Year",
            "Month",
            "Week",
            "MarketType",
            "Species"
        ]
    )


    # ========================================================
    # 13. SORT
    # ========================================================

    month_order = {
        "January": 1,
        "February": 2,
        "March": 3,
        "April": 4,
        "May": 5,
        "June": 6,
        "July": 7,
        "August": 8,
        "September": 9,
        "October": 10,
        "November": 11,
        "December": 12
    }

    df["MonthNumber"] = df["Month"].map(
        month_order
    )

    df = df.sort_values(
        by=[
            "Year",
            "MonthNumber",
            "Week",
            "MarketType",
            "Species"
        ]
    )

    df = df.drop(columns=["MonthNumber"])


    # ========================================================
    # 14. SAVE CSV
    # ========================================================

    df.to_csv(
        output_file,
        index=False,
        encoding="utf-8-sig"
    )


    # ========================================================
    # 15. FINAL REPORT
    # ========================================================

    print("\n")
    print("=" * 80)
    print("SUCCESS")
    print("=" * 80)

    print("Final rows:", len(df))

    print("\nMarket types:")
    print(
        df["MarketType"].value_counts(
            dropna=False
        )
    )

    print("\nUnique species:")
    print(
        df["Species"].nunique()
    )

    print("\nMissing values:")
    print(
        df.isnull().sum()
    )

    print("\nSpecies:")
    print(
        sorted(
            df["Species"]
            .dropna()
            .unique()
        )
    )

    print("\nFirst 20 rows:")
    print(
        df.head(20).to_string(
            index=False
        )
    )

    print("\n")
    print("=" * 80)
    print("CSV SAVED")
    print("=" * 80)

    print(output_file)
```
